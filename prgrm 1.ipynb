```python
a=int(input("Enter 1st no : "))
b=int(input("Enter 2nd no : "))
ch=int(input("Enter your choice: \n 1.addition \n 2.sub \n 3.multi \n 4.div \n choice :"))
if(ch==1):
    print("add:",a+b)
elif(ch==2):
    print("sub:",a-b)
elif(ch==3):
    print("multi:",a*b)
elif(ch==4):
    print( "div:",a/b)
else:
    print("invalid choice")
```
